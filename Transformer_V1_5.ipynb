```python
import torch
import torch.nn as nn
import torch.optim as optim
import math
from collections import Counter
import matplotlib.pyplot as plt
import seaborn as sns

torch.manual_seed(0)  # [NEW] reproducible runs

# ============================================================
# Masks
# ============================================================

def make_padding_mask(seq: torch.Tensor, pad_idx: int = 0) -> torch.Tensor:
    # (batch, seq_len) -> (batch, 1, 1, seq_len); True = keep
    return (seq != pad_idx).unsqueeze(1).unsqueeze(2)

def make_causal_mask(size: int, device=None) -> torch.Tensor:
    return torch.tril(torch.ones((size, size), device=device, dtype=torch.bool)).unsqueeze(0).unsqueeze(0)

def make_tgt_mask(tgt: torch.Tensor, pad_idx: int = 0) -> torch.Tensor:
    return make_padding_mask(tgt, pad_idx) & make_causal_mask(tgt.size(1), tgt.device)


# ============================================================
# Model building blocks
# ============================================================

class PositionalEncoding(nn.Module):
    def __init__(self, d_model: int, max_len: int = 5000, dropout: float = 0.1):
        super().__init__()
        self.dropout = nn.Dropout(dropout)

        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-math.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        self.register_buffer('pe', pe.unsqueeze(0))

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.dropout(x + self.pe[:, :x.size(1)])


class MultiHeadAttention(nn.Module):
    def __init__(self, d_model: int, num_heads: int, dropout: float = 0.1):
        super().__init__()
        assert d_model % num_heads == 0, "d_model must be divisible by num_heads"
        self.d_model = d_model
        self.num_heads = num_heads
        self.d_k = d_model // num_heads

        self.W_q = nn.Linear(d_model, d_model)
        self.W_k = nn.Linear(d_model, d_model)
        self.W_v = nn.Linear(d_model, d_model)
        self.W_o = nn.Linear(d_model, d_model)
        self.attn_dropout = nn.Dropout(dropout)

    def forward(self, q, k, v, mask=None, return_attn: bool = False):
        B = q.size(0)
        Q = self.W_q(q).view(B, -1, self.num_heads, self.d_k).transpose(1, 2)
        K = self.W_k(k).view(B, -1, self.num_heads, self.d_k).transpose(1, 2)
        V = self.W_v(v).view(B, -1, self.num_heads, self.d_k).transpose(1, 2)

        scores = torch.matmul(Q, K.transpose(-2, -1)) / math.sqrt(self.d_k)
        if mask is not None:
            # [NEW] dtype-safe "minus infinity" (-1e9 overflows in float16)
            scores = scores.masked_fill(~mask, torch.finfo(scores.dtype).min)

        attn_weights = torch.softmax(scores, dim=-1)
        context = torch.matmul(self.attn_dropout(attn_weights), V)
        context = context.transpose(1, 2).contiguous().view(B, -1, self.d_model)
        out = self.W_o(context)
        return (out, attn_weights) if return_attn else out


class PositionwiseFeedForward(nn.Module):
    def __init__(self, d_model: int, d_ff: int, dropout: float = 0.1):
        super().__init__()
        self.fc1 = nn.Linear(d_model, d_ff)
        self.fc2 = nn.Linear(d_ff, d_model)
        self.act = nn.GELU()  # [NEW] GELU instead of ReLU (used by BERT/GPT)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        return self.fc2(self.dropout(self.act(self.fc1(x))))


# [NEW] Pre-LayerNorm: normalize BEFORE each sub-layer, keep the residual path clean.
# Trains more stably than the original Post-LN and usually needs less LR tuning.
class EncoderLayer(nn.Module):
    def __init__(self, d_model, num_heads, d_ff, dropout=0.1):
        super().__init__()
        self.self_attn = MultiHeadAttention(d_model, num_heads, dropout)
        self.ffn = PositionwiseFeedForward(d_model, d_ff, dropout)
        self.norm1 = nn.LayerNorm(d_model)
        self.norm2 = nn.LayerNorm(d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x, mask=None):
        h = self.norm1(x)
        attn_out, attn_w = self.self_attn(h, h, h, mask, return_attn=True)
        x = x + self.dropout(attn_out)
        x = x + self.dropout(self.ffn(self.norm2(x)))
        return x, attn_w


class DecoderLayer(nn.Module):
    def __init__(self, d_model, num_heads, d_ff, dropout=0.1):
        super().__init__()
        self.masked_self_attn = MultiHeadAttention(d_model, num_heads, dropout)
        self.cross_attn = MultiHeadAttention(d_model, num_heads, dropout)
        self.ffn = PositionwiseFeedForward(d_model, d_ff, dropout)
        self.norm1 = nn.LayerNorm(d_model)
        self.norm2 = nn.LayerNorm(d_model)
        self.norm3 = nn.LayerNorm(d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x, enc_output, src_mask=None, tgt_mask=None):
        h = self.norm1(x)
        self_out, self_attn_w = self.masked_self_attn(h, h, h, tgt_mask, return_attn=True)
        x = x + self.dropout(self_out)

        h = self.norm2(x)
        cross_out, cross_attn_w = self.cross_attn(h, enc_output, enc_output, src_mask, return_attn=True)
        x = x + self.dropout(cross_out)

        x = x + self.dropout(self.ffn(self.norm3(x)))
        return x, self_attn_w, cross_attn_w


class Transformer(nn.Module):
    def __init__(self, src_size, tgt_size, d_model=512, num_heads=8,
                 num_layers=6, d_ff=2048, max_len=5000, dropout=0.1, pad_idx=0,
                 tie_weights: bool = True):
        super().__init__()
        self.d_model = d_model
        self.pad_idx = pad_idx

        self.src_embedding = nn.Embedding(src_size, d_model, padding_idx=pad_idx)
        # [NEW] Weight tying: same vocab on both sides -> share one embedding matrix,
        # and reuse it for the output projection. Fewer params, better generalization.
        share = tie_weights and src_size == tgt_size
        self.tgt_embedding = self.src_embedding if share else nn.Embedding(tgt_size, d_model, padding_idx=pad_idx)
        self.pos_encoder = PositionalEncoding(d_model, max_len, dropout)

        self.encoder_layers = nn.ModuleList(
            [EncoderLayer(d_model, num_heads, d_ff, dropout) for _ in range(num_layers)])
        self.decoder_layers = nn.ModuleList(
            [DecoderLayer(d_model, num_heads, d_ff, dropout) for _ in range(num_layers)])

        # [NEW] Pre-LN needs a final norm at the end of each stack
        self.enc_norm = nn.LayerNorm(d_model)
        self.dec_norm = nn.LayerNorm(d_model)

        self.linear = nn.Linear(d_model, tgt_size)
        if tie_weights:
            self.linear.weight = self.tgt_embedding.weight

        self._init_parameters()

    def _init_parameters(self):
        # [FIX] The old self.apply(init_weights) re-initialized every parameter once per
        # sub-module, and Xavier overwrote the zero padding row of the embeddings.
        for p in self.parameters():
            if p.dim() > 1:
                nn.init.xavier_uniform_(p)
        with torch.no_grad():
            self.src_embedding.weight[self.pad_idx].zero_()
            self.tgt_embedding.weight[self.pad_idx].zero_()

    def encode(self, src, src_mask=None):
        x = self.pos_encoder(self.src_embedding(src) * math.sqrt(self.d_model))
        for layer in self.encoder_layers:
            x, _ = layer(x, src_mask)
        return self.enc_norm(x)

    def decode(self, tgt, enc_out, src_mask=None, tgt_mask=None, return_attn=False):
        x = self.pos_encoder(self.tgt_embedding(tgt) * math.sqrt(self.d_model))
        cross_attns = []
        for layer in self.decoder_layers:
            x, _, cross_attn_w = layer(x, enc_out, src_mask, tgt_mask)
            cross_attns.append(cross_attn_w)
        logits = self.linear(self.dec_norm(x))
        return (logits, cross_attns) if return_attn else logits

    def forward(self, src, tgt, src_mask=None, tgt_mask=None):
        if src_mask is None:
            src_mask = make_padding_mask(src, self.pad_idx)
        if tgt_mask is None:
            tgt_mask = make_tgt_mask(tgt, self.pad_idx)
        enc_out = self.encode(src, src_mask)
        return self.decode(tgt, enc_out, src_mask, tgt_mask)

    @torch.no_grad()
    def generate(self, src, sos_idx, eos_idx, max_len=20):
        self.eval()
        src_mask = make_padding_mask(src, self.pad_idx)
        enc_out = self.encode(src, src_mask)

        B = src.size(0)
        tgt = torch.full((B, 1), sos_idx, dtype=torch.long, device=src.device)
        finished = torch.zeros(B, dtype=torch.bool, device=src.device)
        last_attn = None

        for _ in range(max_len - 1):
            tgt_mask = make_tgt_mask(tgt, self.pad_idx)
            logits, cross_attns = self.decode(tgt, enc_out, src_mask, tgt_mask, return_attn=True)
            next_token = logits[:, -1, :].argmax(-1, keepdim=True)

            # [FIX] once a sequence has produced <EOS>, pad it instead of
            # letting it keep generating while other batch items finish
            next_token = next_token.masked_fill(finished.unsqueeze(1), self.pad_idx)
            finished |= next_token.squeeze(1) == eos_idx

            tgt = torch.cat([tgt, next_token], dim=1)
            last_attn = cross_attns[-1]
            if finished.all():
                break

        return tgt, last_attn


# ============================================================
# BPE tokenizer
# ============================================================

class BPETokenizer:
    EOW = "</w>"  # [NEW] end-of-word marker so decode() can rebuild real words

    def __init__(self, num_merges: int = 100):
        self.num_merges = num_merges
        self.merges = []
        self.token_to_id = {}
        self.id_to_token = {}
        self.PAD, self.SOS, self.EOS, self.UNK = "<PAD>", "<SOS>", "<EOS>", "<UNK>"

    def _word_to_symbols(self, word):
        # "dog" -> ["d", "o", "g</w>"]
        return list(word[:-1]) + [word[-1] + self.EOW]

    @staticmethod
    def _get_pair_counts(words):
        counts = Counter()
        for symbols in words:
            for i in range(len(symbols) - 1):
                counts[(symbols[i], symbols[i + 1])] += 1
        return counts

    @staticmethod
    def _merge_pair(symbols, pair):
        merged, i = [], 0
        while i < len(symbols):
            if i < len(symbols) - 1 and symbols[i] == pair[0] and symbols[i + 1] == pair[1]:
                merged.append(symbols[i] + symbols[i + 1])
                i += 2
            else:
                merged.append(symbols[i])
                i += 1
        return merged

    def train(self, corpus):
        # [NEW] BPE on unique words weighted by frequency: much faster than
        # merging every occurrence of every word separately
        word_freqs = Counter(w for text in corpus for w in text.lower().split())
        words = {w: self._word_to_symbols(w) for w in word_freqs}

        for _ in range(self.num_merges):
            pair_counts = Counter()
            for w, symbols in words.items():
                for i in range(len(symbols) - 1):
                    pair_counts[(symbols[i], symbols[i + 1])] += word_freqs[w]
            if not pair_counts:
                break
            best_pair, frequency = pair_counts.most_common(1)[0]
            if frequency < 2:
                break
            self.merges.append(best_pair)
            words = {w: self._merge_pair(s, best_pair) for w, s in words.items()}

        symbols = set(s for syms in words.values() for s in syms)
        vocabulary = [self.PAD, self.SOS, self.EOS, self.UNK] + sorted(symbols)
        self.token_to_id = {t: i for i, t in enumerate(vocabulary)}
        self.id_to_token = {i: t for t, i in self.token_to_id.items()}
        self._cache = {}

    def _encode_word(self, word):
        if word in self._cache:  # [NEW] cache: 16k merges per word is slow
            return self._cache[word]
        symbols = self._word_to_symbols(word)
        for pair in self.merges:
            if len(symbols) == 1:
                break
            symbols = self._merge_pair(symbols, pair)
        self._cache[word] = symbols
        return symbols

    def encode(self, text, add_special_tokens=True):
        tokens = [self.SOS] if add_special_tokens else []
        for word in text.lower().split():
            tokens.extend(self._encode_word(word))
        if add_special_tokens:
            tokens.append(self.EOS)
        unk = self.token_to_id[self.UNK]
        return [self.token_to_id.get(t, unk) for t in tokens]

    def decode(self, ids, skip_special_tokens=True):
        specials = {self.PAD, self.SOS, self.EOS}
        tokens = []
        for idx in ids:
            token = self.id_to_token.get(int(idx), self.UNK)
            if skip_special_tokens and token in specials:
                continue
            tokens.append(token)
        # [FIX] join subwords, then use </w> to place the spaces
        return "".join(tokens).replace(self.EOW, " ").strip()

    @property
    def vocab_size(self): return len(self.token_to_id)
    @property
    def pad_id(self): return self.token_to_id[self.PAD]
    @property
    def sos_id(self): return self.token_to_id[self.SOS]
    @property
    def eos_id(self): return self.token_to_id[self.EOS]


# ============================================================
# Attention plot
# ============================================================

def plot_attention_heatmap(src_ids, gen_ids, attention_weights, tokenizer):
    # (batch, heads, tgt_len, src_len) -> average heads
    attn_matrix = attention_weights[0].mean(dim=0).cpu().numpy()

    # [FIX] drop padding columns from the source
    src_ids = [i for i in src_ids if i != tokenizer.pad_id]
    attn_matrix = attn_matrix[:, :len(src_ids)]

    # [FIX] row i of the attention was used to PREDICT gen_ids[i + 1],
    # so the matrix has len(gen_ids) - 1 rows. Label rows with predicted tokens.
    gen_ids = gen_ids[1:1 + attn_matrix.shape[0]]

    label = lambda i: tokenizer.id_to_token.get(int(i), tokenizer.UNK).replace(tokenizer.EOW, "_")
    src_tokens = [label(i) for i in src_ids]
    gen_tokens = [label(i) for i in gen_ids]

    plt.figure(figsize=(10, 8))
    sns.heatmap(attn_matrix, xticklabels=src_tokens, yticklabels=gen_tokens,
                cmap="viridis", annot=False)
    plt.xlabel("Input Sequence (Source)")
    plt.ylabel("Predicted Token (Target)")
    plt.title("Decoder-to-Encoder Cross Attention (last layer, averaged across heads)")
    plt.xticks(rotation=45, ha="right")
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()


# ============================================================
# 1. Data + tokenizer
# ============================================================

data_pairs = [
    ("the dog sat on the floor", "dog resting"),
    ("the cat is sleeping on the sofa", "the cat is sleeping peacefully"),
]

# [FIX] `vocabulary` was never defined in the original script.
# Put your large text block here (or leave empty).
extra_corpus_text = ""

training_corpus = [extra_corpus_text]
for src_text, tgt_text in data_pairs:
    training_corpus += [src_text, tgt_text]

tokenizer = BPETokenizer(num_merges=16000)
tokenizer.train(training_corpus)

VOCAB_SIZE = tokenizer.vocab_size
PAD, SOS, EOS = tokenizer.pad_id, tokenizer.sos_id, tokenizer.eos_id

# ============================================================
# 2. Batching & padding
# ============================================================

src_encoded = [tokenizer.encode(s, add_special_tokens=False) for s, _ in data_pairs]
tgt_encoded = [tokenizer.encode(t, add_special_tokens=True) for _, t in data_pairs]

max_src_len = max(map(len, src_encoded))
max_tgt_len = max(map(len, tgt_encoded))

src = torch.tensor([s + [PAD] * (max_src_len - len(s)) for s in src_encoded], dtype=torch.long)
tgt = torch.tensor([t + [PAD] * (max_tgt_len - len(t)) for t in tgt_encoded], dtype=torch.long)

# ============================================================
# 3. Model
# ============================================================

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = Transformer(
    src_size=VOCAB_SIZE,
    tgt_size=VOCAB_SIZE,
    d_model=512,
    num_heads=4,
    num_layers=2,
    d_ff=2048,
    max_len=100,
    dropout=0.1,
    pad_idx=PAD,
    tie_weights=True,
).to(device)

src, tgt = src.to(device), tgt.to(device)
print(f"Parameters: {sum(p.numel() for p in model.parameters()):,}")

# ============================================================
# 4. Optimizer, scheduler, loss
# ============================================================

EPOCHS = 200
WARMUP_STEPS = 20

# [NEW] AdamW (decoupled weight decay) with the betas from the original paper
optimizer = optim.AdamW(model.parameters(), lr=5e-4, betas=(0.9, 0.98), eps=1e-9, weight_decay=0.01)

# [NEW] Linear warmup, then inverse-sqrt decay (the "Noam" schedule shape)
def lr_lambda(step):
    step = max(step, 1)
    return min(step / WARMUP_STEPS, math.sqrt(WARMUP_STEPS / step))

scheduler = optim.lr_scheduler.LambdaLR(optimizer, lr_lambda)

# [NEW] label smoothing stops the model from becoming over-confident
criterion = nn.CrossEntropyLoss(ignore_index=PAD, label_smoothing=0.1)

# ============================================================
# 5. Training loop
# ============================================================

decoder_input = tgt[:, :-1]  # <SOS> + tokens
expected = tgt[:, 1:]        # tokens + <EOS>

for epoch in range(EPOCHS):
    model.train()
    optimizer.zero_grad()

    logits = model(src, decoder_input)
    loss = criterion(logits.reshape(-1, VOCAB_SIZE), expected.reshape(-1))
    loss.backward()

    # [NEW] gradient clipping guards against loss spikes
    torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
    optimizer.step()
    scheduler.step()

    # [FIX] old check was `epoch % 500` with only 100 epochs -> printed once
    if epoch % 20 == 0 or epoch == EPOCHS - 1:
        # [NEW] token accuracy is easier to read than loss alone
        with torch.no_grad():
            mask = expected != PAD
            acc = (logits.argmax(-1)[mask] == expected[mask]).float().mean().item()
        print(f"Epoch {epoch:4d} | Loss: {loss.item():.4f} | "
              f"Acc: {acc:.2%} | LR: {scheduler.get_last_lr()[0]:.2e}")

# ============================================================
# 6. Generation (all examples)
# ============================================================

generated, attention = model.generate(src, sos_idx=SOS, eos_idx=EOS, max_len=15)

print()
for i, (s, t) in enumerate(data_pairs):
    print(f"Input:    {s}")
    print(f"Expected: {t}")
    print(f"Output:   {tokenizer.decode(generated[i].tolist())}\n")

test_idx = 0
plot_attention_heatmap(
    src_ids=src[test_idx].tolist(),
    gen_ids=generated[test_idx].tolist(),
    attention_weights=attention[test_idx:test_idx + 1],
    tokenizer=tokenizer,
)
```
